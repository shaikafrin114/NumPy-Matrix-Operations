```python
import numpy as np
```

```python
# Creating two 2x2 matrices
matrix_A = np.array([[4, 7],
                     [2, 6]])

matrix_B = np.array([[1, 3],
                     [5, 2]])

print("Matrix A:\n", matrix_A)
print("\nMatrix B:\n", matrix_B)
```

```python
addition_result = matrix_A + matrix_B
print("Addition Result:\n", addition_result)
```

```python
subtraction_result = matrix_A - matrix_B
print("Subtraction Result:\n", subtraction_result)
```

```python
# Using the @ operator as suggested in the mini-guide
multiplication_result = matrix_A @ matrix_B
print("Matrix Multiplication Result:\n", multiplication_result)
```

```python
transpose_A = matrix_A.T
print("Transpose of Matrix A:\n", transpose_A)
```
